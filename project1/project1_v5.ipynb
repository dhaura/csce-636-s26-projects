# Project 1: One Specialist DNN per Parameter Setting

This notebook trains a **separate deep neural network for each of the 9 settings** $(n, k, m)$, with $n = 9$, $k \in \{4, 5, 6\}$ and $m \in \{2, \dots, n-k\}$. Each test sample is sent to the network of its setting, as in a mixture-of-experts model, which the project explicitly allows.

Each network:
- reads the matrix $P$ (zero-padded to $6 \times 5$) and $(n, k, m)$;
- predicts $z = \log_2 \hat y$, bounded to $[0, 100]$; the only post-processing is $\hat y = 2^z$, which is always finite and $\ge 1$;
- is trained in two phases. **Phase 1** uses generated matrices of its setting, whose m-heights were computed exactly. **Phase 2** fine-tunes on the provided training rows of its setting, which follow the distribution of the test set.

A specialist only has to learn one function, so its whole capacity serves its own setting. That matters here: the settings differ in difficulty by a factor of about 100 in squared error, and each setting is graded separately.

The generated training rows come from `generate_fast_data.py` (all settings) and `generate_hard_data.py` (the three settings with $m = n-k$).

- **All settings:** for finite $h_m$, each LP optimum is a vertex $u = s\,G_T^{-1}$ ($k$ columns $T$, signs $s$). So $h_m$ is the largest $\max_t |c_t|$ over vertices whose codeword $c = uG$ has at most $m$ entries above 1.
- **$m = n-k$:** $h_m = \max_{T,\,j \notin T} \lVert G_T^{-1} g_j \rVert_1$.

These algorithms only create training data; the networks never use them.

## Configuration

The network predicts $z=\log_2(\hat y)$, bounded to `[0, 100]`, and $\hat y=2^z$. Each run trains one setting, chosen with `MHEIGHT_SETTING`.

| Setting | Default | Environment variable |
|---|---|---|
| Setting to train (`k-m`, e.g. `6-2`), or `all` for one shared network | `all` | `MHEIGHT_SETTING` |
| Generated datasets added to phase 1 (comma-separated `.npz` files) | all-setting vertex data + closed-form data for $m = n-k$ | `MHEIGHT_EXTRA_DATA` |
| Residual block type / width / blocks | `plain` / 512 / 4 | `MHEIGHT_BLOCK` / `MHEIGHT_WIDTH` / `MHEIGHT_BLOCKS` |
| Optimiser | `adam_plateau` (or `adamw_cosine`) | `MHEIGHT_OPTIMIZER` |
| AdamW weight decay (only with `adamw_cosine`) | 1e-4 | `MHEIGHT_WEIGHT_DECAY` |
| Random row/column permutations and sign flips during training | `0` | `MHEIGHT_AUGMENT` |
| Dropout on each residual branch | 0 | `MHEIGHT_DROPOUT` |
| Phase 1 epochs / learning rate | 20 / 1e-3 | `MHEIGHT_PHASE1_EPOCHS` / `MHEIGHT_LR` |
| Fine-tuning epochs / learning rate | 80 / 3e-4 | `MHEIGHT_FINETUNE_EPOCHS` / `MHEIGHT_FINETUNE_LR` |
| Output directory | `artifacts/dnn_v5_<setting>` | `MHEIGHT_RUN` |

Augmentation and dropout are available but off, because both increased the error in tests. The provided matrices have position-dependent structure that random permutations destroy. Dropout's noise blurs the precise relationships this regression task needs.


```python
import hashlib
import os
import json
import pickle
import numpy as np
from pathlib import Path
import tensorflow as tf

SEED = 636
BATCH_SIZE = 4096
INFERENCE_BATCH_SIZE = 1024
PHASE1_EPOCHS = int(os.environ.get("MHEIGHT_PHASE1_EPOCHS", 20))
FINETUNE_EPOCHS = int(os.environ.get("MHEIGHT_FINETUNE_EPOCHS", 80))
LEARNING_RATE = float(os.environ.get("MHEIGHT_LR", 1e-3))
FINETUNE_LEARNING_RATE = float(os.environ.get("MHEIGHT_FINETUNE_LR", 3e-4))
SETTING = os.environ.get("MHEIGHT_SETTING", "all")
EXTRA_DATA = [path for path in os.environ.get(
    "MHEIGHT_EXTRA_DATA",
    "data/CSCE-636-Project-1-Train-RandomExact.npz,data/CSCE-636-Project-1-Train-RandomExact-MaxM.npz").split(",") if path]
BLOCK = os.environ.get("MHEIGHT_BLOCK", "plain")
OPTIMIZER = os.environ.get("MHEIGHT_OPTIMIZER", "adam_plateau")
WEIGHT_DECAY = float(os.environ.get("MHEIGHT_WEIGHT_DECAY", 1e-4))
AUGMENT = os.environ.get("MHEIGHT_AUGMENT", "0") == "1"
DROPOUT = float(os.environ.get("MHEIGHT_DROPOUT", 0.0))
if BLOCK not in ("gated", "plain") or OPTIMIZER not in ("adamw_cosine", "adam_plateau"):
    raise ValueError("MHEIGHT_BLOCK must be gated/plain and MHEIGHT_OPTIMIZER adamw_cosine/adam_plateau")
MAX_LOG2_HEIGHT = 100.0
WIDTH = int(os.environ.get("MHEIGHT_WIDTH", 512))
NUM_BLOCKS = int(os.environ.get("MHEIGHT_BLOCKS", 4))
ARTIFACT_DIR = Path("artifacts") / os.environ.get("MHEIGHT_RUN", f"dnn_v5_{SETTING}")
ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)
# Model after phase 1 (extended data).
PHASE1_PATH = ARTIFACT_DIR / "phase1_model.keras"
# The fine-tuned model is the final model.
MODEL_PATH = ARTIFACT_DIR / "best_model.keras"
PARAMETER_SETTINGS = [(9, k, m) for k in (4, 5, 6) for m in range(2, 9 - k + 1)]
if SETTING == "all":
    TRAIN_SETTINGS = PARAMETER_SETTINGS
else:
    _k, _m = (int(v) for v in SETTING.split("-"))
    TRAIN_SETTINGS = [(9, _k, _m)]
    if TRAIN_SETTINGS[0] not in PARAMETER_SETTINGS:
        raise ValueError(f"Unknown setting {SETTING}")

# Set TensorFlow to be deterministic and reproducible.
tf.keras.utils.set_random_seed(SEED)

# Enable deterministic operations in TensorFlow to ensure reproducibility.
tf.config.experimental.enable_op_determinism()

# Enable GPU memory growth to avoid allocating all GPU memory at once.
for gpu in tf.config.list_physical_devices("GPU"):
    tf.config.experimental.set_memory_growth(gpu, True)

# Print TensorFlow version and available GPUs.
print("TensorFlow:", tf.__version__, "| Keras:", tf.keras.__version__)
print("GPUs:", tf.config.list_physical_devices("GPU"))
print(f"Run: {ARTIFACT_DIR} | settings: {TRAIN_SETTINGS} | extra data: {EXTRA_DATA} | block: {BLOCK} x {NUM_BLOCKS}, width {WIDTH} | "
      f"optimizer: {OPTIMIZER} | augment: {AUGMENT} | dropout: {DROPOUT}")
```

## Input Formatting and Data Splitting

`P` varies in shape, so we place it in the upper-left corner of a zero-padded `6 x 5` array. The original entries and `[n, k, m]` are the only model inputs. Fixed scaling occurs inside standard Keras `Rescaling` layers.

The same matrix can appear under several `m` values. Hence, a row-wise shuffle could place that matrix in both training and testing. Therefore, a seeded hash of `(n, k, P)` assigns each distinct matrix to training, validation, or testing with probabilities **60%, 20%, 20%**. Identical matrix contents always get the same assignment, even if stored as different NumPy objects. Actual row percentages are approximate because whole matrix groups stay together. The hash is used only for partitioning and is never fed into the model.

Nonfinite or below-one training labels have no valid finite log-loss target. We report and exclude such rows from supervised training and local evaluation.

```python
def pack_features(samples):
    matrices = np.zeros((len(samples), 6, 5), dtype=np.float32)
    parameters = np.empty((len(samples), 3), dtype=np.float32)
    for index, (n, k, m, P) in enumerate(samples):
        if n != 9 or k not in (4, 5, 6) or not 2 <= m <= n - k:
            raise ValueError(f"Invalid parameters at row {index}: {(n, k, m)}")
        if not isinstance(P, np.ndarray) or P.shape != (k, n - k):
            raise ValueError(f"Invalid P shape at row {index}")
        matrices[index, :k, :n - k] = P
        parameters[index] = (n, k, m)
    if not np.isfinite(matrices).all():
        raise ValueError("Input matrices must remain finite after float32 conversion")
    return {"matrix": matrices, "parameters": parameters}


def matrix_partitions(samples, seed):
    partitions = np.empty(len(samples), dtype=np.uint8)
    object_cache = {}
    key = int(seed).to_bytes(8, "little")
    for index, (n, k, m, P) in enumerate(samples):
        cache_key = (int(n), int(k), id(P))
        if cache_key not in object_cache:
            payload = bytes((int(n), int(k))) + np.asarray(P, dtype="<f8").tobytes(order="C")
            digest = hashlib.blake2b(payload, digest_size=8, key=key).digest()
            bucket = int.from_bytes(digest, "little") % 10
            object_cache[cache_key] = 0 if bucket < 6 else (1 if bucket < 8 else 2)
        partitions[index] = object_cache[cache_key]
    return partitions
```

```python
if len(features) != len(labels):
    raise ValueError("Feature and label counts differ")
raw_targets = np.asarray(labels, dtype=np.float64)
if raw_targets.shape != (len(features),):
    raise ValueError("Expected one scalar label per sample")
valid_target = np.isfinite(raw_targets) & (raw_targets >= 1.0)
print(f"Excluded {np.count_nonzero(~valid_target):,} invalid training labels")
if not np.any(valid_target):
    raise ValueError("No valid training labels")

targets_log2 = np.zeros((len(features), 1), dtype=np.float32)
targets_log2[valid_target, 0] = np.log2(raw_targets[valid_target])
if targets_log2[valid_target].max() > MAX_LOG2_HEIGHT:
    raise ValueError("A training target exceeds the model output range")
inputs = pack_features(features)
partition = matrix_partitions(features, SEED)
train_indices = np.flatnonzero(valid_target & (partition == 0))
validation_indices = np.flatnonzero(valid_target & (partition == 1))
test_indices = np.flatnonzero(valid_target & (partition == 2))

setting_ids = np.empty(len(features), dtype=np.int8)
for group_id, (n, k, m) in enumerate(PARAMETER_SETTINGS):
    mask = (inputs["parameters"][:, 1] == k) & (inputs["parameters"][:, 2] == m)
    setting_ids[mask] = group_id

splits = {"train": train_indices, "validation": validation_indices, "test": test_indices}
for name, indices in splits.items():
    if len(indices) == 0:
        raise ValueError(f"The {name} split is empty")
    print(f"{name:10s}: {len(indices):,} rows ({100 * len(indices) / valid_target.sum():.2f}%)")
print("\n(n, k, m)       train       validation       test")
for group_id, setting in enumerate(PARAMETER_SETTINGS):
    counts = [int(np.count_nonzero(setting_ids[indices] == group_id)) for indices in splits.values()]
    if min(counts) == 0:
        raise ValueError(f"Setting {setting} is missing from a split")
    print(f"{str(setting):12s} {counts[0]:10,d} {counts[1]:14,d} {counts[2]:10,d}")
np.savez_compressed(ARTIFACT_DIR / "split_indices.npz", **splits)

# Provided-data rows of each split: used for fine-tuning, model selection and reporting.
provided_train_indices = train_indices[train_indices < N_PROVIDED]
provided_validation_indices = validation_indices[validation_indices < N_PROVIDED]
provided_test_indices = test_indices[test_indices < N_PROVIDED]
print(f"\nProvided-data rows: train {len(provided_train_indices):,}, "
      f"validation {len(provided_validation_indices):,}, test {len(provided_test_indices):,}")
```

## Add the Generated Rows and Select the Setting

The generated `.npz` files contain new matrices, labelled exactly:
- `CSCE-636-Project-1-Train-RandomExact.npz`: all settings, by vertex enumeration (about 40 million rows);
- `CSCE-636-Project-1-Train-RandomExact-MaxM.npz`: the three $m = n-k$ settings, by the closed form (10 million rows each).

Only rows of the trained setting are kept, and they only join **phase 1's training rows**. Validation, fine-tuning and testing use the split defined above.

Then every split (training, validation, test, and their provided-data parts) is restricted to the trained setting, so the network only sees its own setting.


```python
wanted = {(k, m) for _, k, m in TRAIN_SETTINGS}
for extra_path in EXTRA_DATA:
    hard = np.load(extra_path)
    select = np.zeros(len(hard["k"]), dtype=bool)
    for k, m in wanted:
        select |= (hard["k"] == k) & (hard["m"] == m)
    hard = {name: values[select] for name, values in hard.items()}
    hard_heights = hard["height"].astype(np.float64)
    if not np.isfinite(hard_heights).all() or np.any(hard_heights < 1):
        raise ValueError(f"{extra_path}: labels must be finite and at least 1")
    hard_k = hard["k"].astype(np.float32)
    hard_parameters = np.stack([np.full_like(hard_k, 9.0), hard_k, hard["m"].astype(np.float32)], axis=1)
    hard_ids = np.empty(len(hard_k), dtype=setting_ids.dtype)
    for group_id, (n, k, m) in enumerate(PARAMETER_SETTINGS):
        hard_ids[(hard["k"] == k) & (hard["m"] == m)] = group_id
    start = len(raw_targets)
    inputs = {"matrix": np.concatenate([inputs["matrix"], hard["P"].astype(np.float32)]),
              "parameters": np.concatenate([inputs["parameters"], hard_parameters])}
    raw_targets = np.concatenate([raw_targets, hard_heights])
    targets_log2 = np.concatenate([targets_log2, np.log2(hard_heights).astype(np.float32)[:, None]])
    setting_ids = np.concatenate([setting_ids, hard_ids])
    train_indices = np.concatenate([train_indices, np.arange(start, start + len(hard_k))])
    del hard
    print(f"Added {len(hard_k):,} generated rows from {extra_path}")

# Restrict every split to the trained setting(s).
train_ids = [PARAMETER_SETTINGS.index(s) for s in TRAIN_SETTINGS]
def restrict(indices):
    return indices[np.isin(setting_ids[indices], train_ids)]
train_indices, validation_indices, test_indices = map(restrict, (train_indices, validation_indices, test_indices))
provided_train_indices, provided_validation_indices, provided_test_indices = map(
    restrict, (provided_train_indices, provided_validation_indices, provided_test_indices))
counts = np.bincount(setting_ids[train_indices], minlength=len(PARAMETER_SETTINGS))
for setting in TRAIN_SETTINGS:
    print(f"  {setting}: {counts[PARAMETER_SETTINGS.index(setting)]:,} phase 1 training rows")
print(f"Phase 1 rows {len(train_indices):,}; provided rows: train {len(provided_train_indices):,}, "
      f"validation {len(provided_validation_indices):,}, test {len(provided_test_indices):,}")
```

## DNN

The network is a residual multilayer perceptron:

1. **Inputs and fixed scaling:** `P / 100`, flattened from the padded `6 × 5` array, concatenated with `[n, k, m] / 10` into 33 values.
2. **Projection:** `Dense(512)` maps the 33 values to the residual width.
3. **Residual blocks** (4): each computes `x + Dense(512)(Dense(1024, GELU)(LayerNorm(x)))`. The skip connections keep the deep network easy to train.
4. **Output head:** `LayerNorm` and `ReLU`, then `Dense(1)` with Softplus, and a bounded `ReLU(max_value=100)` that gives $z = \log_2 \hat y \in [0, 100]$.

The network has about 4.2 million trainable parameters.

**Optimiser.** Adam with gradient-norm clipping at 1. The learning rate starts at 1e-3 in phase 1 and 3e-4 in fine-tuning, and is halved whenever the validation loss has not improved for 3 epochs. Training stops after 8 epochs without improvement, and the best epoch is restored. With `MHEIGHT_SETTING=all`, rows are weighted inversely by their setting's frequency, so all 9 settings count equally.


```python
def residual_block(x, width, block=BLOCK):
    h = tf.keras.layers.LayerNormalization()(x)
    if block == "gated":
        h = tf.keras.layers.Multiply()([
            tf.keras.layers.Dense(2 * width)(h),
            tf.keras.layers.Dense(2 * width, activation="gelu")(h),
        ])
    else:
        h = tf.keras.layers.Dense(2 * width, activation="gelu")(h)
    h = tf.keras.layers.Dense(width)(h)
    h = tf.keras.layers.Dropout(DROPOUT)(h)
    return tf.keras.layers.Add()([x, h])


def build_model(width=WIDTH, num_blocks=NUM_BLOCKS):
    matrix_input = tf.keras.Input(shape=(6, 5), name="matrix")
    parameter_input = tf.keras.Input(shape=(3,), name="parameters")
    matrix_values = tf.keras.layers.Rescaling(1.0 / 100.0)(matrix_input)
    matrix_values = tf.keras.layers.Flatten()(matrix_values)
    parameter_values = tf.keras.layers.Rescaling(1.0 / 10.0)(parameter_input)
    x = tf.keras.layers.Concatenate()([matrix_values, parameter_values])
    x = tf.keras.layers.Dense(width)(x)
    for _ in range(num_blocks):
        x = residual_block(x, width)
    x = tf.keras.layers.LayerNormalization()(x)
    x = tf.keras.layers.ReLU()(x)
    x = tf.keras.layers.Dense(1, activation="softplus")(x)
    output = tf.keras.layers.ReLU(max_value=MAX_LOG2_HEIGHT, name="log2_height")(x)
    network = tf.keras.Model(
        inputs={"matrix": matrix_input, "parameters": parameter_input},
        outputs=output, name="mheight_resmlp",
    )
    return network


def compile_network(network, learning_rate, epochs, rows):
    if OPTIMIZER == "adamw_cosine":
        steps = epochs * int(np.ceil(rows / BATCH_SIZE))
        warmup = min(2000, steps // 10)
        schedule = tf.keras.optimizers.schedules.CosineDecay(
            initial_learning_rate=learning_rate * 0.01, decay_steps=steps - warmup, alpha=0.01,
            warmup_target=learning_rate, warmup_steps=warmup)
        optimizer = tf.keras.optimizers.AdamW(learning_rate=schedule, weight_decay=WEIGHT_DECAY, clipnorm=1.0)
        optimizer.exclude_from_weight_decay(var_names=["bias", "gamma", "beta"])
    else:
        optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate, clipnorm=1.0)
    network.compile(
        optimizer=optimizer,
        loss="mean_squared_error",
        # Sample weights already balance the loss; no additional metrics requested.
        weighted_metrics=[],
    )

model = build_model()
compile_network(model, LEARNING_RATE, PHASE1_EPOCHS, len(train_indices))
model.summary()
```

```python
def setting_weights(indices):
    counts = np.bincount(setting_ids[indices], minlength=len(PARAMETER_SETTINGS))
    present = [PARAMETER_SETTINGS.index(s) for s in TRAIN_SETTINGS]
    if np.any(counts[present] == 0):
        raise ValueError("Each trained setting must be present in the weighted split")
    return (len(indices) / (len(present) * counts[setting_ids[indices]])).astype(np.float32)


def augment_matrices(features, target, weight):
    """Random permutations and sign flips of P's rows and columns, within its k x (n-k) block.

    Permuting or negating rows of P (with the matching identity columns) or columns of P
    maps the code to an equivalent one, so every m-height is unchanged.
    """
    matrices, parameters = features["matrix"], features["parameters"]
    batch = tf.shape(matrices)[0]
    k = tf.cast(parameters[:, 1], tf.int32)
    width = tf.cast(parameters[:, 0] - parameters[:, 1], tf.int32)
    # Padding rows/columns get keys >= 2, so the sort keeps them at the end.
    row_keys = tf.random.uniform((batch, 6)) + 2.0 * tf.cast(tf.range(6)[None, :] >= k[:, None], tf.float32)
    column_keys = tf.random.uniform((batch, 5)) + 2.0 * tf.cast(tf.range(5)[None, :] >= width[:, None], tf.float32)
    matrices = tf.gather(matrices, tf.argsort(row_keys, axis=1), axis=1, batch_dims=1)
    matrices = tf.gather(matrices, tf.argsort(column_keys, axis=1), axis=2, batch_dims=1)
    row_signs = tf.where(tf.random.uniform((batch, 6, 1)) < 0.5, -1.0, 1.0)
    column_signs = tf.where(tf.random.uniform((batch, 1, 5)) < 0.5, -1.0, 1.0)
    return dict(features, matrix=matrices * row_signs * column_signs), target, weight


def make_dataset(indices, training=False):
    # Keep the dataset on host memory; only batches move to the training device.
    with tf.device("/CPU:0"):
        dataset = tf.data.Dataset.from_tensor_slices((
            {name: values[indices] for name, values in inputs.items()},
            targets_log2[indices], setting_weights(indices),
        ))
    if training:
        dataset = dataset.shuffle(min(len(indices), 100_000), seed=SEED, reshuffle_each_iteration=True)
    options = tf.data.Options()
    options.experimental_deterministic = True
    options.threading.private_threadpool_size = 2
    dataset = dataset.with_options(options).batch(BATCH_SIZE)
    if training and AUGMENT:
        dataset = dataset.map(augment_matrices)
    return dataset.prefetch(1)
```

## Phase 1: Train on the Generated Data

Phase 1 trains on all training rows of the setting: the provided and extended training rows plus the generated rows of that setting. Its validation set is the setting's validation rows. The best weights are restored at the end and saved as `phase1_model.keras`.


```python
def training_callbacks(checkpoint_path):
    return [
        tf.keras.callbacks.ModelCheckpoint(str(checkpoint_path), monitor="val_loss", save_best_only=True),
        tf.keras.callbacks.EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True),
        tf.keras.callbacks.TerminateOnNaN(),
    ] + ([tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-6)]
         if OPTIMIZER == "adam_plateau" else [])


def save_history(history, name):
    if not np.isfinite(history.history["val_loss"]).any():
        raise RuntimeError("Training produced no finite validation loss; do not export predictions")
    with (ARTIFACT_DIR / name).open("w") as file:
        json.dump({key: [float(value) for value in values] for key, values in history.history.items()}, file, indent=2)


phase1_history = model.fit(
    make_dataset(train_indices, training=True),
    validation_data=make_dataset(validation_indices),
    epochs=PHASE1_EPOCHS, callbacks=training_callbacks(PHASE1_PATH),
    verbose=2, shuffle=False,
)
save_history(phase1_history, "phase1_history.json")
print("Best phase 1 validation macro log2 MSE:", min(phase1_history.history["val_loss"]))
print("Saved phase 1 model:", PHASE1_PATH)
```

## Phase 2: Fine-tune on the Provided Data

Fine-tuning starts from the best phase 1 weights. It uses only the provided training rows of the setting, which follow the test-set distribution, with a lower learning rate so the features learned in phase 1 are adjusted rather than overwritten. Early stopping and checkpoint selection use the provided validation rows of the setting.


```python
compile_network(model, FINETUNE_LEARNING_RATE, FINETUNE_EPOCHS, len(provided_train_indices))
finetune_history = model.fit(
    make_dataset(provided_train_indices, training=True),
    validation_data=make_dataset(provided_validation_indices),
    epochs=FINETUNE_EPOCHS, callbacks=training_callbacks(MODEL_PATH),
    verbose=2, shuffle=False,
)
save_history(finetune_history, "finetune_history.json")
print("Best fine-tuning validation macro log2 MSE (provided rows):", min(finetune_history.history["val_loss"]))
print("Saved fine-tuned model:", MODEL_PATH)
```

## Test the DNN

Inference uses the saved networks with `training=False`. The only numerical post-processing is `np.exp2`, which converts $\log_2$ predictions to m-heights.

Both the phase 1 network and the fine-tuned network are evaluated on the provided validation and test rows of the setting. This shows what fine-tuning adds.


```python
def load_prediction_model(model_path=MODEL_PATH):
    with tf.device("/CPU:0"):
        return tf.keras.models.load_model(str(model_path), compile=False)


def model_log2_batch(network, batch):
    with tf.device("/CPU:0"):
        values = np.asarray(network(batch, training=False)).reshape(-1).astype(np.float64)
    if not np.isfinite(values).all() or np.any(values < 0) or np.any(values > MAX_LOG2_HEIGHT):
        raise ValueError("Model returned an invalid log2 m-height; predictions were not saved")
    return values


def predict_heights(network, samples):
    predictions = np.empty(len(samples), dtype=np.float64)
    for start in range(0, len(samples), INFERENCE_BATCH_SIZE):
        stop = min(start + INFERENCE_BATCH_SIZE, len(samples))
        log_values = model_log2_batch(network, pack_features(samples[start:stop]))
        predictions[start:stop] = np.exp2(log_values)
    if not np.isfinite(predictions).all() or np.any(predictions < 1):
        raise ValueError("Predictions must be finite and at least 1")
    return predictions


def save_prediction_list(path, predictions):
    predictions = np.asarray(predictions, dtype=np.float64)
    if predictions.ndim != 1 or not np.isfinite(predictions).all() or np.any(predictions < 1):
        raise ValueError("Expected a one-dimensional list of finite predictions >= 1")
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temporary = path.with_name(path.name + ".tmp")
    with temporary.open("wb") as file:
        pickle.dump(predictions.tolist(), file, protocol=pickle.HIGHEST_PROTOCOL)
    temporary.replace(path)
```

```python
def report_costs(name, indices, inference_model):
    predicted_log2 = np.empty(len(indices), dtype=np.float64)
    for start in range(0, len(indices), INFERENCE_BATCH_SIZE):
        selected = indices[start:start + INFERENCE_BATCH_SIZE]
        batch = {key: values[selected] for key, values in inputs.items()}
        predicted_log2[start:start + len(selected)] = model_log2_batch(inference_model, batch)
    # Use the original float64 labels for reporting the specified cost.
    errors = (np.log2(raw_targets[indices]) - predicted_log2) ** 2
    rows = []
    print(f"\n{name}: per-setting squared log2 error")
    for group_id, setting in enumerate(PARAMETER_SETTINGS):
        mask = setting_ids[indices] == group_id
        if not mask.any():
            continue
        cost = float(errors[mask].mean())
        rows.append({"n": setting[0], "k": setting[1], "m": setting[2],
                     "count": int(mask.sum()), "sigma": cost})
        print(f"{setting}: count={mask.sum():,}, sigma={cost:.6f}")
    macro = float(np.mean([row["sigma"] for row in rows]))
    print(f"Equal-setting mean cost: {macro:.6f}")
    print(f"Row-weighted mean cost: {errors.mean():.6f}")
    return {"per_setting": rows, "macro_sigma": macro, "row_mean_sigma": float(errors.mean())}

phase1_model = load_prediction_model(PHASE1_PATH)
inference_model = load_prediction_model(MODEL_PATH)
cost_report = {
    "phase1": {
        "validation_provided": report_costs("Phase 1, validation (provided rows)", provided_validation_indices, phase1_model),
        "local_test_provided": report_costs("Phase 1, held-out test (provided rows)", provided_test_indices, phase1_model),
    },
    "finetuned": {
        "validation_provided": report_costs("Fine-tuned, validation (provided rows)", provided_validation_indices, inference_model),
        "local_test_provided": report_costs("Fine-tuned, held-out test (provided rows)", provided_test_indices, inference_model),
        "local_test_extended": report_costs("Fine-tuned, held-out test (all extended rows)", test_indices, inference_model),
    },
}
with (ARTIFACT_DIR / "costs.json").open("w") as file:
    json.dump(cost_report, file, indent=2)
```

## Compute σ from Prediction Files

This cell computes the project's cost exactly as the specification defines it, from three lists:
- the input samples `[n, k, m, P]`;
- the true m-heights;
- the predicted m-heights.

For each parameter setting $(n,k,m)$:

$$\sigma_{n,k,m} = \frac{1}{N_{n,k,m}} \sum_t \left(\log_2 y^{[t]} - \log_2 \hat y^{[t]}\right)^2$$

The fine-tuned network's predictions for the provided held-out test rows go through `predict_heights` and `save_prediction_list` (the same path as the final cell), are read back from the pickle, and are then scored.


```python
def compute_sigma(samples, true_heights, predicted_heights):
    true_heights = np.asarray(true_heights, dtype=np.float64)
    predicted_heights = np.asarray(predicted_heights, dtype=np.float64)
    if not len(samples) == len(true_heights) == len(predicted_heights):
        raise ValueError("Samples, true heights and predictions must have the same length")
    if not np.isfinite(predicted_heights).all() or np.any(predicted_heights < 1):
        raise ValueError("Every predicted m-height must be finite and at least 1")
    costs = (np.log2(true_heights) - np.log2(predicted_heights)) ** 2
    settings = [(int(n), int(k), int(m)) for n, k, m, _ in samples]
    sigma = {}
    for setting in PARAMETER_SETTINGS:
        mask = np.array([s == setting for s in settings])
        if mask.any():
            sigma[setting] = float(costs[mask].mean())
            print(f"sigma{setting} = {sigma[setting]:.6f}  (N = {mask.sum():,})")
    print(f"Equal-setting mean sigma: {np.mean(list(sigma.values())):.6f}")
    print(f"Row-weighted mean sigma:  {costs.mean():.6f}")
    return sigma


test_samples = [features[i] for i in provided_test_indices]
test_true_heights = [labels[i] for i in provided_test_indices]
prediction_path = ARTIFACT_DIR / "local_test_predictions.pkl"
save_prediction_list(prediction_path, predict_heights(inference_model, test_samples))
with prediction_path.open("rb") as file:
    test_predictions = pickle.load(file)
test_sigma = compute_sigma(test_samples, test_true_heights, test_predictions)
```

## Final Cell: Predict m-Heights for a Test File

This is the cell for testing the submitted model. It is **self-contained**: in a fresh session, run the first cell (it installs the pinned TensorFlow/Keras versions on Colab), then run this cell. None of the training cells or the training data are needed.

1. Place the 9 specialist models next to this notebook as `best_model_k{k}_m{m}.keras` (for example `best_model_k6_m2.keras`). When run from the project folder, the cell also finds them at `artifacts/dnn_v5_{k}-{m}/best_model.keras`.
2. Set `TEST_INPUT_PATH` to the test file: a pickled list of samples `[n, k, m, P]`, in the same format as `CSCE-636-Project-1-Train-n_k_m_P`.
3. Run the cell. It writes `OUTPUT_PATH`: a pickled list of predicted m-heights (Python floats), one per input sample in the same order, in the same format as `CSCE-636-Project-1-Train-mHeights`.

Each sample goes to the specialist for its $(k, m)$. `P` is placed in a zero-padded `6 × 5` array and `[n, k, m]` is passed alongside it. Each network outputs $z = \log_2 \hat y$ in `[0, 100]`, and the only post-processing is $\hat y = 2^z$. Every prediction is therefore finite and at least 1; the cell checks this before saving. Inference runs on the CPU in fixed batches of 1024, compiled with XLA, so repeated runs on a machine give identical results.


```python
# ===== Final cell: load the trained DNN and predict m-heights for a test file =====
import pickle
from pathlib import Path

import numpy as np
import tensorflow as tf

TEST_INPUT_PATH = "CSCE-636-Project-1-Test-n_k_m_P"   # pickled list of [n, k, m, P]
OUTPUT_PATH = "CSCE-636-Project-1-Test-mHeights"      # pickled list of predicted m-heights
BATCH = 1024


class CompatUnpickler(pickle.Unpickler):
    # Pickles written with NumPy 2 reference numpy._core; NumPy 1.x only has numpy.core.
    def find_class(self, module, name):
        if int(np.__version__.split(".")[0]) < 2 and module.startswith("numpy._core"):
            module = "numpy.core" + module[len("numpy._core"):]
        return super().find_class(module, name)


if not Path(TEST_INPUT_PATH).exists():
    # Nothing to predict yet (e.g. a batch run before the test file is available).
    print(f"Test file {TEST_INPUT_PATH!r} not found: set TEST_INPUT_PATH and run this cell again.")
else:
    def specialist_file(k, m):
        for path in (f"best_model_k{k}_m{m}.keras", f"artifacts/dnn_v5_{k}-{m}/best_model.keras"):
            if Path(path).exists():
                return path
        raise FileNotFoundError(f"No specialist model found for k={k}, m={m}")

    with open(TEST_INPUT_PATH, "rb") as file:
        samples = CompatUnpickler(file).load()

    # Inputs: P in the upper-left corner of a zero-padded 6 x 5 array, and [n, k, m].
    matrices = np.zeros((len(samples), 6, 5), dtype=np.float32)
    parameters = np.zeros((len(samples), 3), dtype=np.float32)
    for row, (n, k, m, P) in enumerate(samples):
        P = np.asarray(P, dtype=np.float64)
        if P.shape != (k, n - k):
            raise ValueError(f"Sample {row}: P has shape {P.shape}, expected {(k, n - k)}")
        matrices[row, :k, :n - k] = P
        parameters[row] = (n, k, m)

    # Each sample goes to the specialist for its (k, m); outputs are z = log2(m-height).
    log2_heights = np.full(len(samples), np.nan)
    used = []
    for k, m in sorted({(int(p[1]), int(p[2])) for p in parameters}):
        rows = np.flatnonzero((parameters[:, 1] == k) & (parameters[:, 2] == m))
        model_file = specialist_file(k, m)
        used.append(model_file)
        with tf.device("/CPU:0"):
            network = tf.keras.models.load_model(model_file, compile=False)

        @tf.function(jit_compile=True)
        def predict_log2(batch):
            return network(batch, training=False)

        with tf.device("/CPU:0"):
            for start in range(0, len(rows), BATCH):
                selected = rows[start:start + BATCH]
                batch = {"matrix": matrices[selected], "parameters": parameters[selected]}
                log2_heights[selected] = np.asarray(predict_log2(batch)).reshape(-1)
    predicted_heights = np.exp2(log2_heights)

    if len(predicted_heights) != len(samples) or not np.isfinite(predicted_heights).all() \
            or np.any(predicted_heights < 1):
        raise ValueError("Every prediction must be finite and at least 1; nothing was saved")
    with open(OUTPUT_PATH, "wb") as file:
        pickle.dump(predicted_heights.tolist(), file)

    print(f"Specialists used: {used} (TensorFlow {tf.__version__}, Keras {tf.keras.__version__})")
    print(f"Predicted {len(predicted_heights):,} m-heights -> {OUTPUT_PATH}")
    print(f"Range: {predicted_heights.min():.4f} to {predicted_heights.max():.4f}; first five: {predicted_heights[:5].tolist()}")
```
